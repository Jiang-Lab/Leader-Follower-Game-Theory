# Fig. 1 — Pairwise payoff structure of the leader–follower game
Schematic of the payoff entries: α (leader→leader), β (followers→leaders), γ (leaders→followers), δ (follower→follower).

```python
import matplotlib.pyplot as plt
from matplotlib.patches import Rectangle, FancyArrowPatch
from lf_model import save
```

```python
fig, ax = plt.subplots(figsize=(18, 7))
ax.set_xlim(0, 18); ax.set_ylim(0, 7); ax.axis("off")
box_lw, arrow_lw, loop_lw, arrow_size, font_box, font_greek = 5, 6, 5.5, 42, 42, 44
leader_x, leader_y, box_w, box_h = 2.0, 2.3, 4.2, 1.8
follower_x, follower_y = 11.8, 2.3

for x0, label in [(leader_x, "Leader"), (follower_x, "Follower")]:
    ax.add_patch(Rectangle((x0, leader_y), box_w, box_h, fill=False, linewidth=box_lw, edgecolor="black"))
    ax.text(x0 + box_w / 2, leader_y + box_h / 2, label, ha="center", va="center",
            fontsize=font_box, fontweight="bold")

# gamma: Leader -> Follower (upper arrow); beta: Follower -> Leader (lower arrow)
ax.add_patch(FancyArrowPatch((leader_x + box_w, leader_y + 1.25), (follower_x, follower_y + 1.25),
             arrowstyle="-|>", mutation_scale=arrow_size, linewidth=arrow_lw, color="black", shrinkA=10, shrinkB=10))
ax.text(9.0, leader_y + 1.85, r"$\gamma$", fontsize=font_greek, ha="center", va="center")
ax.add_patch(FancyArrowPatch((follower_x, follower_y + 0.55), (leader_x + box_w, leader_y + 0.55),
             arrowstyle="-|>", mutation_scale=arrow_size, linewidth=arrow_lw, color="black", shrinkA=10, shrinkB=10))
ax.text(9.0, leader_y - 0.15, r"$\beta$", fontsize=font_greek, ha="center", va="center")

# self-loops alpha (leader) and delta (follower)
ax.add_patch(FancyArrowPatch((leader_x + 1.1, leader_y + box_h + 0.05), (leader_x - 0.35, leader_y + 0.35),
             connectionstyle="arc3,rad=1.05", arrowstyle="-|>", mutation_scale=arrow_size, linewidth=loop_lw, color="black"))
ax.text(leader_x + 0.45, leader_y + box_h + 1.1, r"$\alpha$", fontsize=font_greek, ha="center")
ax.add_patch(FancyArrowPatch((follower_x + box_w - 1.1, follower_y - 0.05), (follower_x + box_w + 0.35, follower_y + 0.85),
             connectionstyle="arc3,rad=1.05", arrowstyle="-|>", mutation_scale=arrow_size, linewidth=loop_lw, color="black"))
ax.text(follower_x + box_w - 0.65, follower_y - 1.1, r"$\delta$", fontsize=font_greek, ha="center")

save(fig, "Fig1_payoff_structure_schematic")
plt.show()
```
