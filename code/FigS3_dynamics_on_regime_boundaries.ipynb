# Fig. S3 — Leader fraction dynamics on the boundaries between game regimes
Growth-coupled model ($N(0)=10^5$, $K=10^{10}$, α = 0.4, δ = 0.6) on the four boundaries
PD–SD (β = δ), PD–CD (γ = α), H–SD (γ = α) and H–CD (β = δ).

```python
import numpy as np
import matplotlib.pyplot as plt
from lf_model import *
set_style()
```

```python
cases = [("A", "PD–SD boundary", 0.60, 0.80), ("B", "PD–CD boundary", 0.20, 0.40),
         ("C", "H–SD boundary", 0.80, 0.40), ("D", "H–CD boundary", 0.60, 0.20)]
x0s = np.array([0.1, 0.3, 0.5, 0.7, 0.9])
colors = ["#b2182b", "#ef8a62", "#cccccc", "#67a9cf", "#2166ac"]
fig, axes = plt.subplots(1, 4, figsize=(17, 4), sharey=True)
for ax, (letter, title, b, gm) in zip(axes, cases):
    _, _, t, xs, _ = simulate(ALPHA, b, gm, DELTA, x0=x0s, T=50, dt=0.01, return_traj=True, n_save=500)
    for k, x0 in enumerate(x0s):
        ax.plot(t, xs[:, k], color=colors[k], lw=2.4, label=f"$x_0={x0}$")
    ax.set_title(title, fontweight="bold"); ax.set_xlabel("Time (days)"); ax.grid(alpha=0.2)
    panel_letter(ax, letter)
    print(f"{title}: x(50) = {np.round(xs[-1], 3)}")
axes[0].set_ylabel("Leader fraction, $x(t)$"); axes[0].set_ylim(-0.02, 1.02)
axes[-1].legend(title="Initial condition", loc="center left", bbox_to_anchor=(1.02, 0.5), frameon=False)
fig.tight_layout()
save(fig, "FigS3_dynamics_on_regime_boundaries")
plt.show()
```
